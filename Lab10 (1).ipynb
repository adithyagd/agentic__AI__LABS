```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    api_key="  ",
    model="llama-3.1-8b-instant",
    temperature=0.3
)

print("Groq Model Connected Successfully")
```

```python
from langchain_community.tools import DuckDuckGoSearchResults

search = DuckDuckGoSearchResults()

def search_agent(topic):
    print("Searching latest news...\n")
    return search.run(f"latest news about {topic}")
```

```python
def summariser_agent(raw_results):

    prompt = f"""
    Summarize the following news into exactly 3 short bullet points.

    News:
    {raw_results}
    """

    response = llm.invoke(prompt)

    return response.content
```

```python
def report_agent(topic, summary):

    prompt = f"""
    Create a Markdown report.

    Title:
    # News: {topic}

    Include:

    - Heading
    - Summary
    - Three bullet points

    Summary:

    {summary}
    """

    response = llm.invoke(prompt)

    return response.content
```

```python
topic = "artificial intelligence"

# Agent 1
raw_results = search_agent(topic)

print("\nSearch Results:")
print(raw_results)

# Agent 2
summary = summariser_agent(raw_results)

print("\nSummary:")
print(summary)

# Agent 3
report = report_agent(topic, summary)

print("\nMarkdown Report:")
print(report)

# Save report
with open("news_report.md", "w", encoding="utf-8") as file:
    file.write(report)

print("\nReport saved successfully as news_report.md")
```
